# Python `heapq` Module: Complete Reference & Practice Guide

The `heapq` module in Python provides an implementation of the **Binary Heap** algorithm based on a regular Python `list`. By default, Python's `heapq` implements a **Min Heap**, where the smallest element is always at the root (`heap[0]`).

### Key Properties of Python Heaps:
* **Zero-based indexing**: For any element at index $k$:
  * Left child is at `2*k + 1`
  * Right child is at `2*k + 2`
  * Parent is at `(k - 1) // 2`
* **Min Heap Invariant**: `heap[k] <= heap[2*k + 1]` and `heap[k] <= heap[2*k + 2]` for all elements where children exist.
* **Storage**: A heap is simply a standard Python `list`. There is no separate `Heap` class; you use `heapq` functions directly on standard lists.

# Python `heapq` Methods

These are the core functions provided by Python's `heapq` module to create, manipulate, and query min-heaps.


### 1. `heapq.heapify()`

#### Description
`heapq.heapify(x)` transforms an arbitrary list `x` into a min-heap **in-place**.

* **Time Complexity**: $\mathcal{O}(n)$, where $n$ is the number of elements in the list.
* **Space Complexity**: $\mathcal{O}(1)$ auxiliary space (in-place modification).
* **Return Value**: Returns `None`. The list is modified directly.
* **Mechanism**: Uses Floyd's heap construction algorithm (sifting down from index $\lfloor n/2 \rfloor - 1$ down to $0$). This is significantly faster than building a heap by calling `heappush()` $n$ times (which would take $\mathcal{O}(n \log n)$).


```python
import heapq

# Arbitrary unsorted list
numbers = [20, 14, 2, 15, 10, 5, 1, 8]
print(f"Original list : {numbers}")

# In-place transformation to min-heap
heapq.heapify(numbers)

print(f"Heapified list: {numbers}")
print(f"Smallest element (root at index 0): {numbers[0]}")

# Verification of min-heap invariant for all parent-child relationships
is_valid_min_heap = all(
    numbers[i] <= numbers[2 * i + 1] and (2 * i + 2 >= len(numbers) or numbers[i] <= numbers[2 * i + 2])
    for i in range(len(numbers) // 2)
)
print(f"Satisfies min-heap property: {is_valid_min_heap}")
```

### 2. `heapq.heappush()`

#### Description
`heapq.heappush(heap, item)` pushes `item` onto the `heap`, maintaining the min-heap invariant.

* **Time Complexity**: $\mathcal{O}(\log n)$ worst-case.
* **Space Complexity**: $\mathcal{O}(1)$ auxiliary space.
* **Mechanism**: Appends `item` to the end of the list and performs a **sift-up** (bubble-up) operation, swapping the item with its parent until the min-heap invariant `parent <= child` is restored.


```python
import heapq

# Starting with an empty heap
task_heap = []

# Pushing elements one by one
items_to_add = [45, 12, 89, 5, 23, 7]
print(f"Adding items sequentially: {items_to_add}\n")

for item in items_to_add:
    heapq.heappush(task_heap, item)
    print(f"Pushed {item:>2} -> Heap: {task_heap} | Current Min: {task_heap[0]}")

print(f"\nFinal min-heap: {task_heap}")
```

### 3. `heapq.heappop()`

#### Description
`heapq.heappop(heap)` pops and returns the smallest item from the `heap`, maintaining the min-heap invariant.

* **Time Complexity**: $\mathcal{O}(\log n)$.
* **Space Complexity**: $\mathcal{O}(1)$.
* **Return Value**: The smallest element in the heap (`heap[0]`).
* **Exceptions**: Raises `IndexError` if `heap` is empty.
* **Mechanism**: 
  1. The minimum element at `heap[0]` is saved to be returned.
  2. The last element in the list is popped and placed at index `0`.
  3. A **sift-down** operation is performed, comparing the element with its smaller child and swapping until the heap invariant is restored.


```python
import heapq

scores = [18, 4, 32, 1, 15, 7]
heapq.heapify(scores)
print(f"Initial heap: {scores}\n")

# Popping elements one-by-one yields elements in sorted ascending order
extracted_elements = []
while scores:
    smallest = heapq.heappop(scores)
    extracted_elements.append(smallest)
    print(f"Popped: {smallest:>2} | Remaining heap: {scores}")

print(f"\nAll elements popped in order: {extracted_elements}")

# Edge case: Popping from an empty heap raises IndexError
try:
    heapq.heappop(scores)
except IndexError as e:
    print(f"Attempting to pop from empty heap caught: {type(e).__name__} ({e})")
```

### 4. `heapq.heappushpop()`

#### Description
`heapq.heappushpop(heap, item)` pushes `item` on the heap, then pops and returns the smallest item from the `heap`.

* **Time Complexity**: $\mathcal{O}(\log n)$.
* **Efficiency**: Combines push and pop into a single fast operation that runs faster than calling `heappush()` followed by `heappop()`.
* **Crucial Behavior**:
  * If the newly pushed `item` is **smaller** than the current root (`heap[0]`), `item` itself is returned immediately, leaving the existing heap completely unchanged!
  * If `item` is greater than or equal to `heap[0]`, `heap[0]` is returned and `item` is sifted into the heap.


```python
import heapq

h = [10, 20, 30, 40, 50]
heapq.heapify(h)
print(f"Original heap: {h} (Current min: {h[0]})\n")

# Case 1: Pushing an item LARGER than current min
pushed_val = 25
result1 = heapq.heappushpop(h, pushed_val)
print(f"Case 1 - Pushed {pushed_val}:")
print(f"  Returned (popped): {result1}")
print(f"  Heap afterwards   : {h}\n")

# Case 2: Pushing an item SMALLER than current min
pushed_val = 5
result2 = heapq.heappushpop(h, pushed_val)
print(f"Case 2 - Pushed {pushed_val} (smaller than root {h[0]}):")
print(f"  Returned (popped): {result2} (item returned directly!)")
print(f"  Heap afterwards   : {h} (heap is unmodified)")
```

### 5. `heapq.heapreplace()`

#### Description
`heapq.heapreplace(heap, item)` pops and returns the smallest item from `heap`, and **then** pushes the new `item`.

* **Time Complexity**: $\mathcal{O}(\log n)$.
* **Heap Size**: Remains unchanged.
* **Return Value**: The smallest element originally in the heap.
* **Exceptions**: Raises `IndexError` if `heap` is empty.
* **Difference from `heappushpop()`**:
  * In `heapreplace()`, the pop occurs **first**. Even if the new `item` is smaller than the original root, the original root is popped, and `item` becomes the new root (or is sifted down).
  * The returned value is **always** an element that was already in the heap, never the newly added `item`.


```python
import heapq

h = [10, 20, 30, 40, 50]
heapq.heapify(h)
print(f"Original heap: {h} (Current min: {h[0]})\n")

# Replace root with a larger number
old_root = heapq.heapreplace(h, 35)
print(f"Replaced root with 35:")
print(f"  Popped old root: {old_root}")
print(f"  New heap       : {h}\n")

# Replace root with a smaller number (e.g., 2)
# Notice: 2 will be placed into the heap, NOT returned!
old_root2 = heapq.heapreplace(h, 2)
print(f"Replaced root with 2 (smaller than current min {old_root2}):")
print(f"  Popped old root: {old_root2} (original element returned)")
print(f"  New heap       : {h} (new element 2 is now in the heap)")
```

### 6. `heapq.nsmallest()`

#### Description
`heapq.nsmallest(n, iterable, key=None)` returns a list with the `n` smallest elements from the dataset defined by `iterable`.

* **Time Complexity**: $\mathcal{O}(m \log n)$, where $m$ is the total number of items in `iterable`.
* **When to use**: Best when $n$ is small compared to $m$. If $n == 1$, `min()` is faster. If $n \approx m$, `sorted(iterable)[:n]` is faster.
* **Parameters**:
  * `n`: Number of smallest elements to retrieve.
  * `iterable`: Any iterable collection.
  * `key`: Optional one-argument function used to extract a comparison key from each element (e.g., `key=lambda x: x['price']`).


```python
import heapq

data = [42, 17, 88, 3, 25, 64, 1, 99, 53, 11]

# Top 4 smallest numbers
smallest_4 = heapq.nsmallest(4, data)
print(f"4 smallest numbers: {smallest_4}")

# Using the key parameter on complex data (list of dictionaries)
products = [
    {"name": "Keyboard", "price": 45, "rating": 4.5},
    {"name": "Mouse",    "price": 25, "rating": 4.2},
    {"name": "Monitor",  "price": 200, "rating": 4.8},
    {"name": "USB Cable","price": 8,   "rating": 4.0},
    {"name": "Webcam",   "price": 60, "rating": 4.3},
]

# Find the 3 cheapest products
cheapest_3 = heapq.nsmallest(3, products, key=lambda item: item["price"])
print("\n3 Cheapest Products:")
for p in cheapest_3:
    print(f"  - {p['name']}: ${p['price']}")
```

### 7. `heapq.nlargest()`

#### Description
`heapq.nlargest(n, iterable, key=None)` returns a list with the `n` largest elements from the dataset defined by `iterable`.

* **Time Complexity**: $\mathcal{O}(m \log n)$, where $m$ is the total number of items in `iterable`.
* **Mechanism**: Maintains a min-heap of size $n$ internally. As elements stream in, any element larger than the min-heap's root replaces the root. The resulting heap contains the top $n$ largest elements, returned sorted in descending order.
* **When to use**: Highly efficient for finding top-$K$ elements in large streams or datasets without sorting the entire dataset.


```python
import heapq

data = [42, 17, 88, 3, 25, 64, 1, 99, 53, 11]

# Top 4 largest numbers (returned in descending order)
largest_4 = heapq.nlargest(4, data)
print(f"4 largest numbers: {largest_4}")

# Using key parameter to find highest-rated products
products = [
    {"name": "Keyboard", "price": 45, "rating": 4.5},
    {"name": "Mouse",    "price": 25, "rating": 4.2},
    {"name": "Monitor",  "price": 200, "rating": 4.8},
    {"name": "USB Cable","price": 8,   "rating": 4.0},
    {"name": "Webcam",   "price": 60, "rating": 4.3},
]

top_rated_2 = heapq.nlargest(2, products, key=lambda item: item["rating"])
print("\nTop 2 Highest Rated Products:")
for p in top_rated_2:
    print(f"  - {p['name']} (Rating: {p['rating']}, Price: ${p['price']})")
```

# Max Heap Methods (Python 3.14+)

> **Note on Python 3.14+**:
> Prior to Python 3.14, Python's `heapq` module only exposed internal/private functions with leading underscores (e.g., `_heapify_max`, `_heappop_max`).
> Starting in **Python 3.14**, max-heap operations are **officially public first-class functions**:
> `heapify_max`, `heappush_max`, `heappop_max`, `heappushpop_max`, and `heapreplace_max`.
>
> In a **Max Heap**:
> * The maximum element is always at the root: `heap[0]`.
> * Max-heap invariant: `heap[k] >= heap[2*k + 1]` and `heap[k] >= heap[2*k + 2]`.


### 8. `heapq.heapify_max()`

#### Description
`heapq.heapify_max(x)` transforms list `x` into a max-heap **in-place**.

* **Time Complexity**: $\mathcal{O}(n)$.
* **Space Complexity**: $\mathcal{O}(1)$ auxiliary space.
* **Return Value**: Returns `None`. Modifies the list in-place.
* **Max Heap Root**: `heap[0]` is guaranteed to be the largest element in the heap.


```python
import heapq

nums = [15, 3, 27, 8, 42, 19, 1]
print(f"Original list : {nums}")

# Transform list to a max-heap in-place
heapq.heapify_max(nums)

print(f"Max-heap list : {nums}")
print(f"Root element (maximum) at index 0: {nums[0]}")

# Verify max-heap property
is_valid_max_heap = all(
    nums[i] >= nums[2 * i + 1] and (2 * i + 2 >= len(nums) or nums[i] >= nums[2 * i + 2])
    for i in range(len(nums) // 2)
)
print(f"Satisfies max-heap property: {is_valid_max_heap}")
```

### 9. `heapq.heappush_max()`

#### Description
`heapq.heappush_max(heap, item)` pushes `item` onto `heap`, maintaining the max-heap invariant.

* **Time Complexity**: $\mathcal{O}(\log n)$.
* **Space Complexity**: $\mathcal{O}(1)$.
* **Mechanism**: Appends `item` to the list and performs a sift-up operation using the maximum comparison (`child <= parent`). If the pushed item is greater than the root, it bubbles all the way up to index `0`.


```python
import heapq

max_heap = []

items = [12, 50, 24, 75, 5, 90]
print(f"Pushing to max-heap sequentially: {items}\n")

for item in items:
    heapq.heappush_max(max_heap, item)
    print(f"Pushed {item:>2} -> Max-Heap: {max_heap} | Current Max: {max_heap[0]}")

print(f"\nFinal max-heap: {max_heap}")
```

### 10. `heapq.heappop_max()`

#### Description
`heapq.heappop_max(heap)` pops and returns the largest element from `heap`, maintaining the max-heap invariant.

* **Time Complexity**: $\mathcal{O}(\log n)$.
* **Return Value**: The largest element in the heap (`heap[0]`).
* **Exceptions**: Raises `IndexError` if the heap is empty.
* **Mechanism**: Swaps `heap[0]` with the last element, pops the largest element, and sifts down the replacement element.


```python
import heapq

values = [30, 80, 10, 95, 45, 60]
heapq.heapify_max(values)
print(f"Initial max-heap: {values}\n")

popped_descending = []
while values:
    largest = heapq.heappop_max(values)
    popped_descending.append(largest)
    print(f"Popped max: {largest:>2} | Remaining max-heap: {values}")

print(f"\nElements retrieved in descending order: {popped_descending}")
```

### 11. `heapq.heappushpop_max()`

#### Description
`heapq.heappushpop_max(heap, item)` pushes `item` on the max-heap, then pops and returns the largest item from the heap.

* **Time Complexity**: $\mathcal{O}(\log n)$.
* **Crucial Behavior**:
  * If `item` is **greater than or equal to** current root `heap[0]`, `item` itself is returned immediately, leaving the max-heap untouched!
  * If `item` is smaller than `heap[0]`, `heap[0]` is popped and returned, and `item` is sifted into the max-heap.


```python
import heapq

max_h = [70, 50, 60, 20, 30]
heapq.heapify_max(max_h)
print(f"Initial max-heap: {max_h} (Current max: {max_h[0]})\n")

# Case 1: Push element smaller than max
res1 = heapq.heappushpop_max(max_h, 40)
print(f"Case 1 - Pushed 40 (smaller than max {70}):")
print(f"  Returned (popped): {res1}")
print(f"  Heap afterwards   : {max_h}\n")

# Case 2: Push element larger than max
res2 = heapq.heappushpop_max(max_h, 99)
print(f"Case 2 - Pushed 99 (larger than current max {max_h[0]}):")
print(f"  Returned (popped): {res2} (item returned directly!)")
print(f"  Heap afterwards   : {max_h} (heap is unmodified)")
```

### 12. `heapq.heapreplace_max()`

#### Description
`heapq.heapreplace_max(heap, item)` pops and returns the largest item from the max-heap first, and **then** pushes `item`.

* **Time Complexity**: $\mathcal{O}(\log n)$.
* **Heap Size**: Remains constant.
* **Return Value**: The original maximum item from the heap (never the newly pushed item).
* **Exceptions**: Raises `IndexError` if the heap is empty.


```python
import heapq

max_h = [80, 60, 70, 20, 40]
heapq.heapify_max(max_h)
print(f"Initial max-heap: {max_h} (Current max: {max_h[0]})\n")

# Replace max with 10
old_max = heapq.heapreplace_max(max_h, 10)
print(f"Replaced max with 10:")
print(f"  Popped maximum: {old_max}")
print(f"  New max-heap  : {max_h} (New max is now {max_h[0]})\n")

# Replace max with an even larger number (e.g. 100)
# Notice: old max is popped and returned; 100 becomes part of heap
old_max2 = heapq.heapreplace_max(max_h, 100)
print(f"Replaced max with 100:")
print(f"  Popped maximum: {old_max2}")
print(f"  New max-heap  : {max_h}")
```
