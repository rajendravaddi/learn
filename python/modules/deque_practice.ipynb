# collections.deque

`deque` (pronounced *"deck"*, short for **double-ended queue**) from the `collections` module is a generalization of stacks and queues. Unlike Python's `list`, which is optimized for fast fixed-length operations and $O(1)$ access at the right end (but $O(n)$ at the left end), `deque` is implemented as a **doubly-linked list of fixed-size blocks**.

### Key Properties of collections.deque:
* **$O(1)$ ends**: Appending and popping from **both** the left and right ends take constant $\mathcal{O}(1)$ time.
* **Memory & Indexing**: Random access (`d[i]`) is $\mathcal{O}(1)$ at the ends, and $\mathcal{O}(n)$ toward the middle.
* **Bounded deques (`maxlen`)**: You can pass `maxlen=N` when creating a `deque`. When full, adding new items to one end automatically discards items from the opposite end.

---
# Part 1: Adding Elements


### 1. append(x)

#### Description
`d.append(x)` adds `x` to the **right end** (tail) of the deque.

* **Time Complexity**: $\mathcal{O}(1)$.
* **Return Value**: `None` (modifies the deque in-place).
* **`maxlen` Behavior**: If the deque is bounded (`maxlen` is set) and already full, an item is automatically discarded from the **left end** to make room.


```python
from collections import deque

d = deque([10, 20, 30])
print(f"Initial deque: {d}")

d.append(40)
d.append(50)
print(f"After append(40) and append(50): {d}")
```

```python

# With maxlen=3: appending to a full deque drops from the left
bounded = deque([1, 2, 3], maxlen=3)
print(f"\nBounded deque (maxlen=3): {bounded}")
bounded.append(4)
print(f"After bounded.append(4):  {bounded} (1 was dropped from left)")
```

### 2. appendleft(x)

#### Description
`d.appendleft(x)` adds `x` to the **left end** (head) of the deque.

* **Time Complexity**: $\mathcal{O}(1)$ (unlike `list.insert(0, x)` which is $\mathcal{O}(n)$).
* **Return Value**: `None` (modifies the deque in-place).
* **`maxlen` Behavior**: If the deque is bounded and full, an item is automatically discarded from the **right end**.


```python
from collections import deque

d = deque([20, 30, 40])
print(f"Initial deque: {d}")

d.appendleft(10)
d.appendleft(0)
print(f"After appendleft(10) and appendleft(0): {d}")
```

```python
# With maxlen=3: appendleft on a full deque drops from the right
bounded = deque([1, 2, 3], maxlen=3)
print(f"\nBounded deque (maxlen=3): {bounded}")
bounded.appendleft(0)
print(f"After bounded.appendleft(0): {bounded} (3 was dropped from right)")
```

### 3. extend(iterable)

#### Description
`d.extend(iterable)` extends the **right end** of the deque by appending all elements from `iterable` in order.

* **Time Complexity**: $\mathcal{O}(k)$, where $k$ is the length of `iterable`.
* **Return Value**: `None` (modifies the deque in-place).
* **Note**: Unlike `append(iterable)` (which adds the iterable as a single nested object), `extend(iterable)` unpacks and appends each item individually.


```python
from collections import deque

d = deque([1, 2])
print(f"Initial deque: {d}")

d.extend([3, 4, 5])
print(f"After extend([3, 4, 5]): {d}")

# Works with any iterable (strings, tuples, generators, etc.)
d.extend("AB")
print(f"After extend('AB'): {d}")
```

### 4. extendleft(iterable)

#### Description
`d.extendleft(iterable)` extends the **left end** of the deque by repeatedly calling `appendleft()` for each element in `iterable`.

* **Time Complexity**: $\mathcal{O}(k)$, where $k$ is the length of `iterable`.
* **Return Value**: `None` (modifies the deque in-place).
* **Important Nuance**: Because elements are added one by one to the left end, the series of left appends **reverses the order** of elements in `iterable`.
  * For example, `extendleft([1, 2, 3])` pushes `1`, then `2` in front of `1`, then `3` in front of `2` — resulting in `3, 2, 1` at the left end.


```python
from collections import deque

d = deque([10, 20])
print(f"Initial deque: {d}")

# 1 is pushed left, then 2, then 3 -> appears as 3, 2, 1 at the front
d.extendleft([1, 2, 3])
print(f"After extendleft([1, 2, 3]): {d}")
```

```python

# Same behavior with strings
letters = deque(["X", "Y"])
letters.extendleft("ABC")
print(f"letters after extendleft('ABC'): {letters}")
```

### 5. insert(i, x)

#### Description
`d.insert(i, x)` inserts `x` into the deque at position `i`.

* **Time Complexity**: $\mathcal{O}(n)$ worst-case (since elements must be shifted).
* **Return Value**: `None` (modifies the deque in-place).
* **Index Clamping**: Negative indices count from the right end. Indices out of bounds are clamped to `0` (beginning) or `len(d)` (end).
* **Exceptions**: If a bounded deque (`maxlen`) is already at full capacity, `insert()` raises an `IndexError` rather than silently discarding elements.


```python
from collections import deque

d = deque(["a", "b", "d", "e"])
print(f"Initial deque: {d}")

d.insert(2, "c")
print(f"After insert(2, 'c'):  {d}")

# Negative index inserts relative to the end
d.insert(-1, "before_last")
print(f"After insert(-1, ...): {d}")
```

```python

# Bounded deque at capacity raises IndexError on insert
bounded = deque([1, 2, 3], maxlen=3)
try:
    bounded.insert(1, 99)
except IndexError as e:
    print(f"\nInsert on full bounded deque raised: {type(e).__name__} ({e})")
```

# Part 2: Removing Elements


### 6. pop()

#### Description
`d.pop()` removes and returns an element from the **right end** of the deque.

* **Time Complexity**: $\mathcal{O}(1)$.
* **Return Value**: The rightmost element.
* **Exceptions**: Raises `IndexError` if no elements are present (empty deque).
* **Note**: Unlike `list.pop(i)`, `deque.pop()` does **not** accept an index argument.


```python
from collections import deque

d = deque([10, 20, 30, 40])
print(f"Initial deque: {d}")

right_item = d.pop()
print(f"Popped from right: {right_item}")
print(f"Deque after pop(): {d}")
```

```python

# Empty deque check
empty_d = deque()
try:
    empty_d.pop()
except IndexError as e:
    print(f"\nPopping empty deque raised: {type(e).__name__} ({e})")
```

### 7. popleft()

#### Description
`d.popleft()` removes and returns an element from the **left end** of the deque.

* **Time Complexity**: $\mathcal{O}(1)$ (unlike `list.pop(0)` which is $\mathcal{O}(n)$).
* **Return Value**: The leftmost element.
* **Exceptions**: Raises `IndexError` if the deque is empty.
* **Use Case**: Ideal for FIFO (First-In, First-Out) queues alongside `append()`.


```python
from collections import deque

# FIFO Queue simulation
queue = deque(["task_1", "task_2", "task_3"])
print(f"Initial queue: {queue}\n")

while queue:
    current = queue.popleft()
    print(f"Processing '{current}' | Remaining queue: {queue}")
```

### 8. remove(x)

#### Description
`d.remove(x)` removes the **first occurrence** of `x` (searching from left to right).

* **Time Complexity**: $\mathcal{O}(n)$.
* **Return Value**: `None` (modifies the deque in-place).
* **Exceptions**: Raises `ValueError` if `x` is not found in the deque.


```python
from collections import deque

d = deque([10, 20, 30, 20, 40])
print(f"Initial deque: {d}")

# Removes only the first occurrence of 20
d.remove(20)
print(f"After remove(20): {d}")

# Attempting to remove a missing element raises ValueError
try:
    d.remove(999)
except ValueError as e:
    print(f"\nremove(999) raised: {type(e).__name__} ({e})")
```

### 9. clear()

#### Description
`d.clear()` removes all elements from the deque, leaving it with length `0`.

* **Time Complexity**: $\mathcal{O}(n)$.
* **Return Value**: `None` (modifies the deque in-place).
* **Note**: Preserves any `maxlen` setting on the deque.


```python
from collections import deque

d = deque([1, 2, 3, 4, 5], maxlen=5)
print(f"Before clear(): {d} | len={len(d)}, maxlen={d.maxlen}")

d.clear()
print(f"After clear():  {d} | len={len(d)}, maxlen={d.maxlen}")
```

# Part 3: Rearranging and Inspecting


### 10. rotate(n)

#### Description
`d.rotate(n=1)` rotates the deque `n` steps to the **right**. If `n` is negative, it rotates to the **left**.

* **Time Complexity**: $\mathcal{O}(k)$, where $k = \min(|n|, \text{len}(d))$ (effectively $\mathcal{O}(1)$ for small fixed rotations).
* **Return Value**: `None` (modifies the deque in-place).
* **Equivalence**:
  * When `d` is non-empty, `d.rotate(1)` is equivalent to `d.appendleft(d.pop())`.
  * `d.rotate(-1)` is equivalent to `d.append(d.popleft())`.


```python
from collections import deque

d = deque([1, 2, 3, 4, 5])
print(f"Original:       {d}")

# Rotate right by 2 steps
d.rotate(2)
print(f"rotate(2):      {d}  (last 2 items moved to front)")

# Rotate left by 2 steps (back to original)
d.rotate(-2)
print(f"rotate(-2):     {d}  (first 2 items moved to back)")

# Rotate left by 1 step (default n is 1 when no argument is passed)
d.rotate()
print(f"rotate() [n=1]: {d}")
```

### 11. reverse()

#### Description
`d.reverse()` reverses the elements of the deque **in-place**.

* **Time Complexity**: $\mathcal{O}(n)$.
* **Return Value**: `None` (modifies the deque in-place).
* **Note**: Similar to `list.reverse()`. If you want a reversed iterator without modifying `d`, use `reversed(d)`.


```python
from collections import deque

d = deque(["first", "second", "third", "fourth"])
print(f"Before reverse(): {d}")

result = d.reverse()
print(f"After reverse():  {d}")
print(f"Return value:     {result}")
```

### 12. count(x)

#### Description
`d.count(x)` counts the number of deque elements equal to `x`.

* **Time Complexity**: $\mathcal{O}(n)$.
* **Return Value**: An integer (`0` if `x` is not present).


```python
from collections import deque

d = deque(["a", "b", "a", "c", "a", "b"])
print(f"Deque: {d}")

print(f"count('a'): {d.count('a')}")
print(f"count('b'): {d.count('b')}")
print(f"count('z'): {d.count('z')}")
```

### 13. index(x)

#### Description
`d.index(x[, start[, stop]])` returns the zero-based position of the first match of `x` in the deque (at or after index `start` and before index `stop`).

* **Time Complexity**: $\mathcal{O}(n)$.
* **Return Value**: Integer index of the first matching element.
* **Exceptions**: Raises `ValueError` if `x` is not found in the specified range.


```python
from collections import deque

d = deque([10, 20, 30, 20, 50])
print(f"Deque: {d}")

print(f"index(20):       {d.index(20)}")
print(f"index(20, 2):    {d.index(20, 2)}  (searching from index 2 onwards)")
print(f"index(50, 2, 5): {d.index(50, 2, 5)}")

try:
    d.index(99)
except ValueError as e:
    print(f"\nindex(99) raised: {type(e).__name__} ({e})")
```

### 14. copy()

#### Description
`d.copy()` creates and returns a **shallow copy** of the deque (preserving `maxlen` if set).

* **Time Complexity**: $\mathcal{O}(n)$.
* **Return Value**: A new `deque` object containing references to the same elements.
* **Shallow vs Deep**: Adding or removing items in the copy does not affect the original deque, **but** mutating a mutable element inside it (like a nested list) affects both deques.


```python
from collections import deque

original = deque([[1, 2], [3, 4]], maxlen=4)
cloned = original.copy()

print(f"Original: {original} | maxlen={original.maxlen}")
print(f"Cloned:   {cloned} | maxlen={cloned.maxlen}")
print(f"Same object (is)?: {original is cloned}")

# Structural change on cloned does not affect original
cloned.append([5, 6])
print(f"\nAfter cloned.append([5, 6]):")
print(f"  Original: {original}")
print(f"  Cloned:   {cloned}")

# Mutating a shared nested list affects both (shallow copy)
cloned[0].append(99)
print(f"\nAfter cloned[0].append(99) (shallow copy effect):")
print(f"  Original: {original}")
print(f"  Cloned:   {cloned}")
```
